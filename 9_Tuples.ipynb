```python
# The immutability caveat

data = (10, [1, 2, 3])
data[1].append(4)   # Modifies the list inside the tuple
print(data)
```

```python
# Creating tuples 

tuple1 = (2, 4, 5, 6, 7, 2)
tuple2 = (4, "hi", 6, "me", 78)
```

```python
# Accessing items in a tuple

tuple1 = (2, 4, 5, 6, 7, 2)
tuple1[1]
```

```python
# Access multiple items in a tuple

tuple1 = (2, 4, 5, 6, 7, 2)
tuple1[1:4]
```

```python
# Access all the items

tuple1 = (2, 4, 5, 6, 7, 2)
tuple1
```

```python
# Access items beginning from index 0

tuple1 = (2, 4, 5, 6, 7, 2)
tuple1[:5]
```

```python
# Access items from index 2 till the end

tuple1 = (2, 4, 5, 6, 7, 2)
tuple1[2:]
```

```python
# Slicing as a form of accessing data

tuple1 = (2, 4, 5, 6, 7, 2)
tuple1[1:3]
```

```python
# Deleting tuple 1

del tuple1
```

```python
# Updating tuples

# Convert the tuple to a list
t = (1, 2, 3)
temp = list(t)
temp[0] = 99
t = tuple(temp)
print(t)

# Concatenation
t2 = t + (4,)
print(t2)
```

```python
# Modifying a list inside a tuple

my_tuple = ("Alice", 25, ["apples", "bananas"])

my_tuple[2].append("cherries")
my_tuple[2][0] = "oranges"

print(my_tuple)
```
