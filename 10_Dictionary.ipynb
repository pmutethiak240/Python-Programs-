```python
# Dictionary example

dict1 = {
    'name': 'xyz',
    'age': 25,
    'hobby': 'Dancing'
}
```

```python
# Accessing data from dictionary

dict1 = {
    'name': 'xyz',
    'age': 25,
    'hobby': 'Dancing'
}

dict1['age']

print(dict1['name'])
print(dict1)
```

```python
# Updating data in dictionary

dict1 = {
    'name': 'xyz',
    'age': 25,
    'hobby': 'Dancing'
}

dict1['name'] = 'abc'

dict1 = {
    'name': 'xyz',
    'age': 25,
    'hobby': 'Dancing'
    'Profession': 'pilot'
}
```

```python
# Deleting data in a dictionary

del dict1['name']

del dict1

dict1.clear()
```

```python
# Using keyword arguments

user = dict(name = "Alex", age = 30, role = "Data Analyst", is_active = True)

print (user)
```
