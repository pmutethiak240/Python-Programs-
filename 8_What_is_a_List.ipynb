```python
# Creating a list
fruits = ["apple", "banana", "cherry"]

# Accessings items (0-indexed)
print(fruits[0])    # Output: 'apple'

# Modifying items (mutable)
fruits[1] = "blueberry"     # Changes 'banana' to 'blueberry'

# Adding items
fruits.append("dragonfruit")
```

INDEXING / SLICING (ACCESSING DATA FROM LISTS)

```python
fruits = ["apple", "banana", "cherry", "date", "elderberry"]

# Positive indices - left to right
print(fruits[0])    # Outputs first item - apple

# Negative indices - right to left
print(fruits[-2])   # Outputs second to last item - date
```

```python
# Slicing illustration

numbers = [0, 10, 20, 30, 40, 50, 60]

print(numbers[1:4]) # Output indices 1, 2, 3 - 10, 20, 30
```

4 WAYS TO DELETE DATA FROM LISTS

```python
# 1) Using 'del' - by index / slice
colors = ["red", "green", "blue", "yellow", "purple"]
del colors[0]   # Removes 'red'
del colors[1:3] # Removes 'blue' and 'yellow'
print(colors)   # Output ["green", "purple"]

# 2) Using '.pop()' - removes and returns item
tasks = ["email", "coding", "meeting"]
current_task = tasks.pop(0) # Removes and returns 'email'
print(current_task)
last_task = tasks.pop()     # Defaults to removing last item - 'meeting' 
print(tasks)    # Output ["coding"]

# 3) Using '.remove()' - by value
pets = ["cat", "dog", "bird", "dog"]
pets.remove("dog")  # Removes ONLY the first "dog"
print(pets)     # Output: ['cat', 'bird', 'dog']

# 4) Using '.clear()' - empty the entire list
items = [1, 2, 3]
items.clear()
print(items)    # Output []
```
