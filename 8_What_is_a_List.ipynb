```python
# Creating a list
fruits = ["apple", "banana", "cherry"]

# Accessings items (0-indexed)
print(fruits[0])    # Output: 'apple'

# Modifying items (mutable)
fruits[1] = "blueberry"     # Changes 'banana' to 'blueberry'

# Adding items
fruits.append("dragonfruit")
```
