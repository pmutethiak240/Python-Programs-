```python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

```python
# Prints 0, 1, 2, 3, 4
for i in range(5):
    print(i)
```

```python
# range(start, stop, step)
for i in range(0, 11, 2):
    print(i)
```

```python
# counts down from 5 to 1
for i in range(5, 0, -1):
    print(i)
```

```python
for num in range(1,10,2):
    print(num)
```

```python
count = 0

while count <= 20:
    print(count)
    count += 5  # This is your step! It changes the index by 5 every time
```

```python
num = 0
while (num < 10):
    print(num)
    num = num + 1   # update the value of num by 1
```

```python
count = 1
while count <= 3:
    print(f"Count is {count}")
    count += 1  # Crucial step:
# Increments count so the condition eventually becomes False
```
