```python
# A typical if statement looks like
if condition:
    statements true
```

```python
if 5==5:
    print("You successfully learned if statement.")
```

```python
if 5>9:
    print("Oops! Not this time.")
```

```python
# An if-else statement looks like
if condition:
    statements true
else:
    statements false
```

```python
if 5==3:
    print("You successfully learned if statement.")
else:
    print("Wow! You also learned about else statement.")
```

```python
# elif statements
if condition1:
    statements
elif condition2:
    statements
elif condition3:
    statements
else:
    statements
```

```python
if 5==4:
    print("An if statement. Oh!")
elif 4==4:
    print("That's something new.")
elif 4>=9:
    print("Really?")
else:
    print("Not this time")
```

```python
if 14==14:
    print("I love Coding")
elif 41==4:
    print("I am Learning how to code")
elif 4>=9:
    print("What is Python")
else:
    print("Not this time")
```
